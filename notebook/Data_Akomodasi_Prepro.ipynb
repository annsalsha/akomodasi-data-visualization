# Cleaning

lowercase

```python
cols = ["Kategori", "Nama_Tempat", "Alamat", "Fasilitas", "Keunggulan"]

for col in cols:
    df[col] = df[col].astype(str).str.lower()
```

menghapus spasi dan newline di awal/akhir pada 3 kolom

```python
cols = ['Nama_Tempat', 'Fasilitas', 'Keunggulan']

for col in cols:
    df[col] = df[col].astype(str).str.strip()  # hapus spasi dan newline di awal/akhir
```

membersihan dan membuat konsistensi pada kolom no. telp

```python
import re
import pandas as pd
import numpy as np

def clean_phone_numbers(phone):
    if pd.isna(phone):
        return phone  # biarkan NaN tetap NaN

    # Pisahkan nomor kalau ada koma
    numbers = [n.strip() for n in phone.split(',')]

    cleaned_numbers = []
    for num in numbers:
        # Hapus semua karakter non-digit
        num = re.sub(r'\D', '', num)

        # Kalau diawali '62', ganti jadi '0'
        if num.startswith('62'):
            num = '0' + num[2:]

        # Kalau belum diawali '0' dan tidak kosong, tambahkan '0'
        if num != '' and not num.startswith('0'):
            num = '0' + num

        cleaned_numbers.append(num)

    # Gabungkan kembali pakai koma
    return ', '.join(cleaned_numbers)


df['No_Telp'] = df['No_Telp'].apply(clean_phone_numbers)
```

mengubah singkatan jl

```python
import pandas as pd
import re

def clean_jalan(text):
    if pd.isna(text):
        return text
    # Ganti jl/jl. jadi jalan
    text = re.sub(r'\bjln?\.?\b', 'jalan', text)
    # Hapus titik setelah kata jalan
    text = re.sub(r'jalan\.', 'jalan', text)
    # Hapus spasi ganda kalau ada
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['Alamat'] = df['Alamat'].apply(clean_jalan)
```

menghapus karakter selain angka (1500000\u2063)

```python
import pandas as pd
import re

# Fungsi untuk hapus semua karakter non-digit
def clean_non_digit(text):
    if pd.isna(text):
        return text
    return re.sub(r'[^\d\-\s]', '', text)  # hanya simpan angka, spasi, dan strip

# Terapkan ke kolom harga
df['Harga'] = df['Harga'].apply(clean_non_digit)
```

memisahkan kolom harga menjadi 2 kolom dan membetulkan format

```python
import pandas as pd
import re

def split_harga_safe(x):
    if pd.isna(x) or str(x).strip() == '':
        return pd.Series([pd.NA, pd.NA])

    # Bersihkan karakter non-digit kecuali strip
    x = re.sub(r'[^\d\-\s]', '', str(x))

    # Split berdasarkan strip
    parts = x.split('-')

    min_harga = parts[0].strip()
    max_harga = parts[1].strip() if len(parts) > 1 else ''

    # Convert ke int, kalau kosong tetap pd.NA
    min_harga = int(min_harga) if min_harga != '' else pd.NA
    max_harga = int(max_harga) if max_harga != '' else pd.NA

    # Normalisasi harga kurang dari 1000 → kali 1000
    if min_harga is not pd.NA and min_harga < 1000:
        min_harga *= 1000
    if max_harga is not pd.NA and max_harga < 1000:
        max_harga *= 1000

    return pd.Series([min_harga, max_harga])

# Terapkan ke DataFrame, pakai Int64 untuk mendukung missing value
df[['Harga_min', 'Harga_max']] = df['Harga'].apply(split_harga_safe).astype('Int64')
```

hapus kolom harga dan memindahkan kolom harga min dan max kedepan

```python
import pandas as pd

# Hapus kolom harga
df = df.drop(columns=['Harga'])

# Tentukan urutan kolom baru
cols = list(df.columns)
# Ambil index kolom No_telp
idx = cols.index('No_Telp')
# Susun ulang kolom: semua kolom sampai No_telp + harga_min/harga_max + sisa kolom
new_cols = cols[:idx+1] + ['Harga_min', 'Harga_max'] + [c for c in cols[idx+1:] if c not in ['Harga_min','Harga_max']]

df = df[new_cols]
```

membuat point untuk kolom keunggulan

```python
import pandas as pd

def ubah_jadi_point(teks):
    # Biarkan null tetap null
    if pd.isna(teks) or str(teks).strip() == '':
        return pd.NA

    # Kalau sudah ada strip di awal teks, kembalikan apa adanya
    if str(teks).strip().startswith('-'):
        return teks

    # Ganti newline dengan spasi
    teks = teks.replace('\n', ' ')

    # Pisah berdasarkan titik
    bagian = teks.split('.')

    # Hapus spasi awal/akhir dan bagian kosong
    bagian = [b.strip() for b in bagian if b.strip()]

    # Tambahkan tanda '-'
    return '\n'.join(['- ' + b for b in bagian])

df['Keunggulan'] = df['Keunggulan'].apply(ubah_jadi_point)
```

membuat kolom baru untuk mengetahui tersedia di platform mana

```python
import pandas as pd

# Kolom-kolom platform
platform_cols = ['Traveloka', 'Tiket', 'Booking', 'Agoda', 'Trip', 'OYO', 'Reddoorz']

# Buat kolom baru, kosong jadi NaN
def gabung_platform(row):
    hasil = ', '.join([col for col in platform_cols if row[col] == 1])
    return hasil if hasil else np.nan  # kosong jadi NaN

df['Platform_tersedia'] = df[platform_cols].apply(gabung_platform, axis=1)
```
